```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score

data = ''
df = pd.read(data)
y_true = data['Class']
```

```python
model = np.zeroes(len(y_true))
accuracy = accuracy_score(y_true, model)
print(accuracy)
```

Despite the model simply assuming that none of the transactions were fraudulent, it had outstanding accuracy. This illustrates why accuracy is not always the best score. This is an example of both class imbalance and asymetric costs, so the auc score may also not be a good fit.
